```python
# Tell Me Again:

n = int(input())
current_term = "1"
for _ in range(n):
    print(" ".join(current_term))
    next_term = ""
    count = 1
    for j in range(1, len(current_term)):
        if current_term[j] == current_term[j - 1]:
            count += 1
        else:
            next_term += str(count) + current_term[j - 1]
            count = 1
    next_term += str(count) + current_term[-1]
    current_term = next_term
```

```python
# harry to hogwarts:

n = int(input())
steps = 0
while n != 6174:
    digits = f"{n:04d}"
    x = int("".join(sorted(digits)))
    y = int("".join(sorted(digits, reverse=True)))
    n = y - x
    steps += 1
print(steps)
```

```python
# prime number (total number of divisor of a number is prime or not):

n = int(input())
count = 0
for i in range(1, int(n**0.5) + 1):
    if n % i == 0:
        count += 1
        if i != n // i:
            count += 1
is_prime = True
if count <= 1:
    is_prime = False
for i in range(2, int(count**0.5) + 1):
    if count % i == 0:
        is_prime = False
        break
print(is_prime)

# t = int(input())
# for case in range(t):
#     n = int(input())
#     count = 0
#     for i in range(1, int(n**0.5) + 1):
#         if n % i == 0:
#             count += 1
#             if i != n // i:
#                 count += 1
#     is_prime = True
#     if count <= 1:
#         is_prime = False
#     for i in range(2, int(count**0.5) + 1):
#         if count % i == 0:
#             is_prime = False
#             break
#     print(is_prime)
```

```python
# Write a program for the given integers P and X, that display the positive integer pairs (i, j) with satisfy the following conditions.

i + j == P
i XOR j == X

P = int(input())
X = int(input())

for i in range(P + 1):
    j = P - i
    if i >= 0 and j >= 0 and (i ^ j) == X and i <= j:
        print(i, j)
```
