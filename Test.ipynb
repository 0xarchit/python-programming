```python
# Given an integer N. Compute the sum of all natural numbers from 1 to N . If less than 1 the output is 0. In python
N = int(input())
if N < 1:
    print(0)
else:
    print(N * (N + 1) // 2)
```

```python
# write a program that takes a string contaning only paranthesis and astriks and print True if paranthesis are balanced and false otherwise
'''
example 1
input: "((*)"
output: true
Example 2
input: "*(*)"
output: True
Example 3
input: ())
output: False
'''

stack = []
s = input("")
for char in s:
    if char == '(':
        stack.append(char)
    elif char == ')':
        if stack and stack[-1] == '(':
            stack.pop()
        elif stack and stack[-1] == '*':
            stack.pop()
        else:
            stack.append(char)
    elif char == '*':
        if stack and stack[-1] == '(':
            stack.pop()
        elif stack and stack[-1] == '*':
            stack.pop()
        else:
            stack.append(char)
if stack and stack[-1] == '*':
    stack.pop()
print(not stack)
```
