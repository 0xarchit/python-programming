```python
'''
scope of variable
1. local
2. global
3. non local
'''
```

```python
def add():
    a = 10 #local
    print(a)
a = 100 #global
add()
print(a)
```

```python
def add():
    print(a)
    a = 10 #local
    # error: UnboundLocalError 
a = 100 #global
add()
print(a)
```

```python
def add():
    global a
    a = 10
    print(a)
a = 100 #global
add()
print(a)
```

```python
def add():
    def xyz():
        a = 23
    def non_xyz():
        a = 34
    def gla_xyz():
        a = 400
    a = 10
    xyz()
    print(a)
    non_xyz()
    print(a)
    gla_xyz()
    print(a)
a = 100 #global
add()
print(a)
```

```python
def add():
    def xyz():
        a = 23
    def non_xyz():
        nonlocal a
        a = 34
    def gla_xyz():
        
        a = 400
    a = 10
    xyz()
    print(a)
    non_xyz()
    print(a)
    gla_xyz()
    print(a)
a = 100 #global
add()
print(a)
```

```python
def add():
    def xyz():
        a = 23
    def non_xyz():
        nonlocal a
        a = 34
    def gla_xyz():
        global a
        a = 400
    a = 10
    xyz()
    print(a)
    non_xyz()
    print(a)
    gla_xyz()
    print(a)
a = 100 #global
add()
print(a)
```
