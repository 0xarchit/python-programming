```python
r = 5
h = 10
v = 3.14 * r**2 * h
t = int(input("enter time in minutes: "))
flow_rate = 15
volume_flow = flow_rate * t
if volume_flow > v:
    print("Overflow, water in tank:", volume_flow-v, "m^3")
elif volume_flow < v:
 ;   remaining_height = (v - volume_flow) / (3.14 * r**2)
    print("Underflow, Remaining height of water level:", remaining_height "meters")
else:
    print("Full", v, "m^3")
```

```python
r = 5
h = 10
vtank = 3.14 * r**2 * h
t = float(input("enter time in minutes: "))
f=15
vwtr=f*t
print('user given time', t, 'minutes')
if vwtr==vtank:
    print('tank is full')
elif vwtr>vtank:
    print('tank is overflow')
    print('overflow volume:', vwtr-vtank, 'm^3')
else:
    print('underflow condition')
    ht= vwtr/(3.14*r**2)
    hr= h-ht
    print('filled height', ht, 'm')
    print('remaining height', hr, 'm')
```
