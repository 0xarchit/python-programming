```python
# burjh khalifa
height = int(input())
max_width = len(bin(height - 1)[2:]) 
for i in range(height + 1):
   binary_number = bin(i)[2:]
   print(binary_number.rjust(max_width))
```

```python
# pattern t23 m
n = int(input())
total_rows = 2 * n
for i in range(n):
    stars = '*' * (n - i)
    spaces = ' ' * (2 * i)
    print(stars + spaces + stars)
for i in range(n):
    stars = '*' * (i + 1)
    spaces = ' ' * (2 * (n - i - 1))
    print(stars + spaces + stars)
```

```python
# word representation
num = int(input())  
words = ""  
ones = ["", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]  
tens = ["", "", "twenty", "thirty", "forty", "fifty", "sixty", "seventy", "eighty", "ninety"]  
teens = ["ten", "eleven", "twelve", "thirteen", "fourteen", "fifteen", "sixteen", "seventeen", "eighteen", "nineteen"]  
if num == 0:  
    words = "zero"  
elif num>= 1000:  
    words += ones[num // 1000] + " thousand "  
    num %= 1000  
if num>= 100:  
    words += ones[num // 100] + " hundred "  
    num %= 100  
if num>= 10 and num<= 19:  
    words += teens[num - 10] + " "  
    num = 0  
elif num>= 20:  
    words += tens[num // 10] + " "  
    num %= 10  
if num>= 1 and num<= 9:  
    words += ones[num] + " "  
print(words.strip())
```

```python
# right angle triangle
n = int(input())
num = 1
for i in range(1, n+1):
    for j in range(1, i+1):
        print(num, end=" ")
        num += 1
    print()
```

```python
#A shape

rows = int(input())
position = int(input())
for i in range(1, rows+1):
    for j in range(rows-i):
        print(" ", end="")
    for j in range(2*i-1):
        if j == 0 or j == 2*i-2 or i == position:
            print('*', end='')
        else:
            print(' ', end='')
    print()
```
