```python
# Matrix operations
'''
m1
2 3 4
4 5 6

m2
1 1 6
8 1 2
'''
m1 = [[2, 3, 4], [4, 5, 6]]
m2 = [[1, 1, 6], [8, 1, 2]]
r1 = len(m1)
c1 = len(m1[0])

r2 = len(m2)
c2 = len(m2[0])

if (r1, c1 == r2, c2):
    # m = ([[0]*c1]*r1).copy()
    m = eval(str([[0]*c1]*r1))
    for i in range(r1):
        for j in range(c1):
            m[i][j] = m1[i][j] + m2[i][j]
    print('\n matrix m1')
    for i in m1:
        print(*i, sep='\t')
    print('\n matrix m2')
    for i in m2:
        print(*i, sep='\t')
    print("\n addition of m1 and m2")
    for i in m:
        print(*i, sep='\t')
else:
    print("invalid dimensions for addition")
```

```python
# multiplication of matrix
m1 = [[2, 3, 4], [4, 5, 6]]
m2 = [[1, 1, 6], [8, 1, 2]]
r1 = len(m1)
c1 = len(m1[0])
r2 = len(m2)
c2 = len(m2[0])

if (r1, c1 == r2, c2):
    # m = ([[0]*c1]*r1).copy()
    m = eval(str([[0]*c1]*r1))
    for i in range(len(m1)):
        for j in range(len(m2[0])):
            for k in range(len(m2)):
                m[i][j] += m1[i][k] * m2[k][j]
    print(m)
else:
    print("invalid dimensions for addition")
```

```python
# input matrix as an array
'''
4 3
10 25 6 20 30 40 50 60 70 80 90 100 110 120
'''
def display_mat(m):
    for row in m:
        print(*row, sep="\t")
r, c = list(map(int, input().split()))
raw_lst = list(map(int, input().split()))
m = []
count = 0
if r * c == len(raw_lst):
    tmp = []
    for i in raw_lst:
        tmp.append(i)
        count += 1
        if count == c:
            m.append(tmp)
            tmp = []
            count = 0
display_mat(m)
```
