```python
# 5. write a python program whether given number is perfect or not  (number= divisor)
n= int(input('enter a number to check: '))
sum = 0
for i in range(1,n):
    if n % i == 0:
        sum += i
if n==sum:
        print("perfect")
else:
        print("not perfect")
```

```python
# show all perfect number between 1 to 1000
for num in range(1, 1001):
    sum = 0
    for i in range(1, num):
        if num%i==0:
            sum += i
    if sum == num:
        print(num)  
```

```python
# 8. write python program to display all prime number between a give range
start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))
for i in range(start, end + 1):
    if i > 1:
        for j in range(2, i):
            if (i % j) == 0:
                break
        else:
            print(i, end=',')
```

```python
# parenthesis balacning
input_string = input("Enter a string of parentheses and asterisks: ")
open_count = 0
wildcard_count = 0
for char in input_string:
    if char == '(':
        open_count += 1
    elif char == ')':
        if open_count > 0:
            open_count -= 1
        else:
            if wildcard_count > 0:
                wildcard_count -= 1
            else:
                print("The parentheses are not balanced.")
                break
    elif char == '*':
        wildcard_count += 1
if open_count == 0:
    print("The parentheses are balanced.")
else:
    if open_count <= wildcard_count:
        print("The parentheses are balanced.")
    else:
        print("The parentheses are not balanced.")
```
