```python
n = int(input())

current_term = "1"
result = []

for i in range(n):
    result.append(current_term)
    next_term = " "
    count = 1
    
    for j in range(1, len(current_term)):
        if current_term[j] == current_term[j - 1]:
            count += 1
        else:
            next_term += str(count) + current_term[j - 1]
            count = 1
    
    next_term += str(count) + current_term[-1]
    current_term = next_term

print("\n".join(result))
```

```python
t = int(input())
for i in range(t):
    n = int(input())
    count = 0
    for i in range(1, int(n**0.5) + 1):
        if n % i == 0:
            count += 1
            if i != n // i:
                count += 1
    is_prime = True
    if count <= 1:
        is_prime = False
    for i in range(2, int(count**0.5) + 1):
        if count % i == 0:
            is_prime = False
            break
    print(is_prime)
```

```python
rows = int(input())
if rows <= 1 or rows >= 100:
    raise ValueError("Number of rows must be between 1 and 100.")

# Top part of the hourglass
for i in range(rows):
    # Print leading spaces
    print(" " * i, end="")
    if i % 2 == 0:
        # Print digits from 0 to 9
        for j in range(9 - i):
            print((j % 10), end="")
    else:
        # Print alternating 9s and 0s
        for j in range(9 - i):
            print((9 if j % 2 == 0 else 0), end="")
    print()

# Bottom part of the hourglass
for i in range(rows):
    # Print leading spaces
    print(" " * (rows - i - 1), end="")
    if i % 2 == 0:
        # Print digits from 0 to 9
        for j in range(5 + i):
            print((j % 10), end="")
    else:
        # Print alternating 9s and 0s
        for j in range(5 + i):
            print((9 if j % 2 == 0 else 0), end="")
    print()
```
