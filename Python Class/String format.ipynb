```python
# %format
a = 10
b = 20
c = a + b
tmp = 'sum of %d and  %d is %d' % (a, b, c) # %d, %s, %f, %t
print(tmp)
```

```python
# .format
a = 10
b = 20
c = a + b
tmp = 'sum of {} and  {} is {}'.format(a, b, c)
print(tmp)
```

```python
#f tag formatting
a = 23
b = 20
out = a + b
tmp= f'sum of {a} and {b} is {out}'
print(tmp)
```

```python
#f tag formatting
a = 23
b = 100000
out = a + b
tmp = f'''
value: {a:10}
valu2: {b:10}
'''
print(tmp)
```

```python
#f tag formatting
a = 23
b = 100000
out = a + b
tmp = f'''
value: {a:<10}
valu2: {b:<10}
'''
print(tmp)
```

```python
#f tag formatting
a = 10.0
b = 1.023232
out = a + b
tmp = f'''
value: {a:20.6f}
valu2: {b:20.6f}
'''
print(tmp)
```
