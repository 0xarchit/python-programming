```python
# write python program to find length(or number of digit) of the user given integer number using loop
num = int(input("Enter a number: "))
a= num
count = 0
while num != 0:
    num = num // 10
    count += 1
print(f"Number of digits in the number in {a} is: {count}")
```

```python
# reversing of given number using while looping (input 1234 output 4321)
num = int(input("Enter a number: "))
reversed_num = 0
while num > 0:
    remainder = num % 10
    reversed_num = (reversed_num * 10) + remainder
    num = num // 10
print("Reversed number is: ", reversed_num)
```

```python
# to check given number is armstrong or not using while loop
num = int(input("Enter a positive integer number: "))
order = 0
sum = 0
temp = num
while num:
    num //= 10
    order += 1
num = temp
while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10

if num == sum:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```
