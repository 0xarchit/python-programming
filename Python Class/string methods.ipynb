```python
# captiialize()
st = 'python programming'
print(st.capitalize())
```

```python
#center(width, fill char)
st = 'Hello'
print(st.center(20, ' '))
print(st.center(2, ' '))
```

```python
#ljust()
print(st.ljust(10, '@'))
```

```python
#rjust()
print(st.rjust(10, '@'))
```

```python
#zfill(width) char is bydefault 0
st = '+2132342'
print(st.zfill(20))
```

```python
#upper()/lower()/casefold() 
# lower(): only ascii
# casefold(): all character, like alpha beta, etc
st = 'heLlo 2 World'
print('\n', st.upper(), '\n', st.lower(), '\n', st.casefold())
```

```python
'''strip(), lstrip(), rstrip()'''
st = '\n\thello\n\t'
print(st.strip())
print(st.rstrip())
print(st.lstrip())
```

```python
#split(): split the string with any pattern
st = 'hello python programmingg'
print(st.split())
print(st.split('ll'))
print(st.split('o'))
print(st.split('g'))
print(st.split('gg'))
```

```python
# split in user data
data = '12 45 129 9'
data2 = data.split()
print(data2)
```
