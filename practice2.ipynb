```python
# even odd game
# sample inp : 1010 sample out: false
# difference of odd places digit sum and even places digit sum is odd then true else false
n = int(input())
def evnodd(a):
    a = str(a)
    odd_sum = 0
    even_sum = 0
    for i in range(len(a)):
        if i % 2 == 0:
            even_sum += int(a[i])
        else:
            odd_sum += int(a[i])
    diff = even_sum - odd_sum
    return False if diff % 2 == 0 else True
print(evnodd(n))
```

```python
# program takes two strings and merges them by taking characters alternatively from each string if one string is longer than other remaining character from longer string are appended at end of merged string.
# sample input: python, rocks output: proycthkons
def merge_strings(str1, str2):
    result = ""
    min_len = min(len(str1), len(str2))
    for i in range(min_len):
        result += str1[i] + str2[i]
    result += str1[min_len:] + str2[min_len:]
    return result

print(merge_strings("jai", "gla"))
```

```python
# armstrong number
n = int(input())
l = len(str(n))
sum = 0
original_n = n
for i in range(l):
    digit = n % 10
    sum += digit ** l
    n //= 10
if sum == original_n:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
# 2nd highest frequency in list
def fr(lst):
    if len(lst) < 2:
        return None
    freq = {}
    for item in lst:
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1
    # print(freq)
    # print(type(freq))
    freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)
    if len(freq) < 2:
        return None
    # print(type(freq))
    return freq[1][0]
print(fr([1,23,4,1,2,3,1,3,2,1,1,2]))
```

```python
# parenthesis bnalancing
stack = []
s = input()
for char in s:
    if char == '(':
        stack.append(char)
    elif char == ')':
        if stack and stack[-1] == '(':
            stack.pop()
        elif stack and stack[-1] == '*':
            stack.pop()
        else:
            stack.append(char)
    elif char == '*':
        if stack and stack[-1] == '(':
            stack.pop()
        elif stack and stack[-1] == '*':
            stack.pop()
        else:
            stack.append(char)
if stack and stack[-1] == '*':
    stack.pop()
print(not stack)
```
