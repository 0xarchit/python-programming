```python
def add(a,b):
    return a+b
print(add(10,20))
```

```python
add = lambda a,b: a+b
print(add(10,20))
```

```python
name = ["John", "Alice", "Bob", "Eve", "Mike"]
name.sort(key=len)
print(name)
```

```python
name = ["John", "Alice", "Bob", "Eve", "Mike"]
name.sort(key=lambda x: x[-1])
print(name)
```

```python
name = ["John", "Alice", "Bob", "Eve", "Mike"]
name.sort(key=lambda x: sum([ord(i) for i in x]))
print(name)
```

```python
ls = [2, 4, 6, 7]
print(list(map(lambda x: x + 2, ls)))
```

```python
name = ["john", "Alice", "bob", "Eve", "Mike"]
print(tuple(map(lambda x: x.capitalize() , name)))
```

```python
arr = list(map(int, input().split()))
n = len(arr)
tot_sum = sum(arr)
x = tot_sum // n
if x * n == tot_sum:
    print(x)
else:
    print(-1)
```
