```python
'''
11111
11111
11111
11111
11111
Number pattern 1
'''
for i in range(5):
    for j in range(5):
        print('1', end='')
    print()
```

```python
'''
11111
00000
11111
00000
11111
Number pattern 2
'''
for i in range(6):
    for j in range(5):
        if i % 2 == 0:
            print('1', end='')
        else:
            print('0', end='')
    print()
```
