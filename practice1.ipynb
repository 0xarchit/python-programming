```python
# program to subtract without minus operator
def subtract(x, y):
    while y != 0:
        borrow = (~x) & y
        x = x ^ y
        y = borrow << 1
    return x
subtract(20,10)
```

```python
# find hcf of two numbers
def hcf(a, b):
    while b:
        a, b = b, a % b
    return a
print(hcf(15,20))
```

```python
# print all prime number from 1 to n
def print_primes(n):
    for num in range(1, n + 1):
        if num > 1:
            is_prime = True
            for i in range(2, num):
                if num % i == 0:
                    is_prime = False
                    break
            if is_prime:
                print(num)
print_primes(20)
```

```python
# find sum  of natural number / factorial of number of all natural number from 1 to n.
n = int(input())
sum = 0
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
for i in range(1, n + 1):
    sum += i / factorial(i)
print(f"{sum:.4f}")
```

```python
"""Finds the maximum difference between two numbers in a list.

  Args:
    numbers: A list of integers.

  Returns:
    The maximum difference between any two numbers in the list.
  """
def max_diff(numbers):

  max_diff = 0
  for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
      diff = abs(numbers[i] - numbers[j])
      if diff > max_diff:
        max_diff = diff
  return max_diff

t = int(input())
for i in range(t):
  n = int(input())
  numbers = list(map(int, input().split()))
  max_diff_value = max_diff(numbers)
  print(max_diff_value)
```

```python
# minimum difference
def min_diff(numbers):
    min_diff = float('inf')
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            diff = abs(numbers[i] - numbers[j])
            if diff < min_diff:
                min_diff = diff
    return min_diff

t = int(input())
for i in range(t):
    n = int(input())
    numbers = list(map(int, input().split()))
    min_diff_value = min_diff(numbers)
    print(min_diff_value)
```

```python
# traverse array
# find smallest and largest elements from one dimensional array elements
def find_smallest_largest(arr):
    smallest = largest = arr[0]
    for i in arr:
        if i < smallest:
            smallest = i
        elif i > largest:
            largest = i
    return smallest, largest

arr = list(map(int, input().split()))
smallest, largest = find_smallest_largest(arr)
print(smallest)
print(largest)
```

```python
# remove prime
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

def remove_primes(lst):
    return [num for num in lst if not is_prime(num)]

lst = list(map(int, input().split(',')))
result = remove_primes(lst)
print(result)
```
